STUDENT DATA ANALYZER USING LIST COMPREHENSION

```python
students = ['Achyuth', 'Sai', 'Shivaprasad', 'arjun','SURYA']
marks = [85, 95, 72, 63, 23]


passed_marks = [x for x in marks if x >= 40]
print("passed_marks:", passed_marks)


failed_marks = [x for x in marks if x < 40]
print("failed_marks:", failed_marks)


top_marks = [x for x in marks if x > 80]
print("top_marks:", top_marks)


bonus_marks = [x + 5 for x in marks]
print("bonus_marks:", bonus_marks)


passed_students = [
    y for y, mark in zip(students, marks)
    if mark >= 40
]
print("passed_students:", passed_students)


failed_students = [
    y for y, mark in zip(students, marks)
    if mark < 40
]
print("failed_students:", failed_students)


upper_names = [y.upper() for y in students]
print("upper_names:", upper_names)


long_names = [y for y in students if len(y) > 4]
print("long_names:", long_names)


student_result = [
    f"{student} - {mark}"
    for student, mark in zip(students, marks)
]
print("student_result:", student_result)
```
