```python
class EncTest:

    def __init__(self, a):
        self.__a = a

    def geta(self):
        return self.__a

    def seta(self, a):
        self.__a = a
```

```python

e = EncTest(10)
```

```python

print(e.geta())
```

```python

e.seta(50)

print(e.geta())
```

### Access Modifiers

```python
class employee:
    
    def __init__(self, na, sal):
        self.name=na  #Public Attributes
        self.salary=sal # Public Attributes
        
e1=employee("Sachin",10000)
print(e1.name)
print(e1.salary)

e1.salary=20000
print(e1.salary)
print(e1.name)
```

```python
class employee:
    def __init__(self, name, sal):
        self._name=name  # protected attribute 
        self._salary=sal # protected attribute

e1=employee("Sachin", 10000)
print(e1._name)
print(e1._salary)

e1._salary=20000
e1._name="Sourab"
print(e1._name)
print(e1._salary)
```

```python
class employee:
    def __init__(self, name, sal):
        self.__name=name  # private attribute 
        self.__salary=sal # private attribute

e1=employee("Sachin",10000)

print(e1._employee__salary)
print(e1._employee__name)

e1._employee__salary=20000
e1._employee__name="Sourab"

print(e1._employee__salary)
print(e1._employee__name)
```

```python
class Employee:
   empCount = 0

   def __init__(self, name, salary):
      
      self.name = name
      self.salary = salary
       
      Employee.empCount += 1
   
   def displayCount(self):
     print ("Total Employee %d" % Employee.empCount)

   def displayEmployee(self):
      print ("Name : ", self.name,  ", Salary: ", self.salary)

emp1 = Employee("Sachin", 10000)
emp2 = Employee("Sourab", 20000)

emp1.displayEmployee()  
emp2.displayEmployee()  

print ("Total Employee %d" % Employee.empCount) 
```

```python
class Employee:

    empCount = 0

    def __init__(self, name, salary):
        self.name = name
        self.salary = salary
        Employee.empCount += 1
   
    def displayCount(self):
        print ("Total Employee %d" % Employee.empCount)

    def displayEmployee(self):
        print ("Name : ", self.name,  ", Salary: ", self.salary)

emp1 = Employee("Sachin", 10000)
emp2 = Employee("Sourab", 20000)

print ("Employee.__doc__:", Employee.__doc__)

print ("Employee.__name__:", Employee.__name__)

print ("Employee.__module__:", Employee.__module__)

print ("Employee.__bases__:", Employee.__bases__)

print ("Employee.__dict__:", Employee.__dict__ )
```

```python
class Addition:
    
    def __init__(self, a, b):
        self.a = a
        self.b = b
        print ("In Constructor")
    
    def add(self):
        print(self.a + self.b)
    
    def __del__(self):
        class_name = self.__class__.__name__
        print (class_name, "destroyed")

add_obj = Addition(3,4)
```
