### Bank Application 

```python

class Bank:
    def __init__(self, name,bal):
        self.user_name = name
        self.balance = bal
   
    def show_balance(self):
        print(f"{self.user_name} has balanceRs.{self.balance} ")
        
    def withdraw_money(self, amount):
        if amount > self.balance:
            print("You don't have sufficient funds")
        else:
            self.balance -= amount
            print(f"{self.user_name} withdrawn Rs.{self.balance}")
    
    def deposit_money(self, amount):
            self.balance += amount
            print(f"{self.user_name} deposited  Rs.{self.balance} ")
```

```python
acc = Bank("XYZ Bank",0.0)
```

```python
acc
```

```python

acc.deposit_money(1000)
```

```python

acc.show_balance()
```

```python
acc.withdraw_money(500)
```

```python
acc.show_balance()
```

### Shopping Cart

```python
class ShoppingCart:

    def __init__(self, customer, total):
        self.customer = customer
        self.total = total

    def show_total(self):
        print(f"{self.customer} cart total is Rs.{self.total}")

    def add_item(self, price):
        self.total += price
        print(f"Item added. Cart total is Rs.{self.total}")

    def remove_item(self, price):
        if price > self.total:
            print("Item price is greater than cart total")
        else:
            self.total -= price
            print(f"Item removed. Cart total is Rs.{self.total}")

    def checkout(self):
        print(f"{self.customer} paid Rs.{self.total}")
        self.total = 0
        print("Cart is now empty")


cart = ShoppingCart("Ravi", 0)

cart
```

```python
cart.add_item(500)
```

```python
cart.add_item(300)
```

```python
cart.show_total()
```

```python
cart.remove_item(300)
```

```python
cart.checkout()
```

```python
cart.show_total()
```

### sending an object as an argument to another class

```python
class Student:
       def __init__(self, name):
              self.name = name

class Person:
    def __init__(self, student):
              self.student = student
              
    def showStudent(self):
              print("Student Name: ",self.student.name)
              
    def getStudent(self):
              return self.student

s1 = Student("Ram")

p1 = Person(s1)
p1.showStudent()

old_s1 = p1.getStudent()
print("Student Name : ",old_s1.name)
```
