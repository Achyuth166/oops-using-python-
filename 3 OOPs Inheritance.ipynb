```python
class Student:
    
    def __init__(self, Number, Name):
        self.number = Number
        self.name = Name
    
    def displaystud(self):
        print(f"{self.number} is the regd. number of student '{self.name}'")


class Marks(Student):

    def __init__(self, Number, Name, Test1, Test2):
    
        super().__init__(Number, Name)
        
        self.test1 = Test1
        self.test2 = Test2
    
    def displaymarks(self):
        print(f"{self.test1} and '{self.test2}' are test scores")

stud_marks =  Marks(100,"Sachin", 88, 99)

stud_marks.displaystud()

stud_marks.displaymarks()
```

```python
class Book:
    
    def __init__(self, author, title):
        self.author = author
        self.title = title
        
    def book_info(self):
        print(f"{self.title} is authored by {self.author}")

class Fiction(Book):
    
    def __init__(self, author, title, publisher):
        
        super().__init__(author, title)
        self.publisher = publisher

    def book_info(self):
        
        super().book_info()
        print(f"{self.title} by {self.author} published {self.publisher}")
    
book1 = Fiction("Daniel Silva", "Prince of Fire", "Berkley")

book1.book_info()
```

### Method Overloading and Overriding 

```python
def f(n):
    return n + 10
 
def f(n,m):
    return n + m + 10

print(f(3, 4))
```

```python
class Addition:  
    def Sum(self,a,b):  
        return a+b;  
    
class Multiplication:  
    def Mul(self,a,b):  
        return a*b;  

class Derived(Addition,Multiplication):  
    def Divide(self,a,b):  
        return a/b;  

d = Derived()  
print(d.Sum(10,20))  
print(d.Mul(10,20))  
print(d.Divide(20,10)) 
```

#### Abstract Method

```python
from abc import ABC, abstractmethod

class Computer(ABC):
    @abstractmethod
    def process(self):
        pass
```

```python
class Laptop(Computer):
     def process(self):
        print("working on Laptop")
        
class Desktop(Computer):
    def process(self):
        print("working on Desktop")
```

```python
class Employee:
   
    def setData(self,id,name,gender):
        self.__id=id
        self.__name=name
        self.__gender=gender
    def showData(self):
        print("Id: ",self.__id)
        print("Name: ", self.__name)
        print("Gender: ", self.__gender)

class BankEmployee(Employee):
  
    def setEmployee(self,id,name,gender,comp,dept):
        self.setData(id,name,gender)
        self.__company=comp
        self.__dept=dept
    def showEmployee(self):
        self.showData()
        print("Bank: ", self.__company)
        print("Department: ", self.__dept)

class SoftwareEmployee(Employee): 
    
    def setEmployee(self,id,name,gender,comp,dept):
        self.setData(id,name,gender)
        self.__company=comp
        self.__dept=dept
    def showEmployee(self):
        self.showData()
        print("Softwre: ", self.__company)
        print("Department: ", self.__dept)

def main():
    print("Bank Employee Object")
    e1=BankEmployee()
    e1.setEmployee(100,"Sachin","Male","OBC","Manager")
    e1.showEmployee()
    
    print("\nSoftware Employee")
    e2 = SoftwareEmployee()
    e2.setEmployee(101, "Sourab", "Male", "Info Solutions", "Data Science")
    e2.showEmployee()

if __name__=="__main__":
    main()
```
